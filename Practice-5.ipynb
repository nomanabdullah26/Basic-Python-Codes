```python
#wap write the factorial of first n numbers (using for  loop )

n=5
fact=1

for i in range(1, n+1):
    fact *=i

print("factorial number is : " , fact)
```

```python
#wap write the factorial of first n numbers (using for  loop )

n=5
fact=1
i=1

while i<=n:
  fact *=i
  i +=1
print("factorial number is : " , fact)
```
