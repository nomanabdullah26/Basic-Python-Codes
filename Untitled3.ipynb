```python
# ============================================================
# Figure Generation for UAV-CAS TGNN Paper
# ============================================================
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from matplotlib.patches import Rectangle
import os

# Output directory
os.makedirs("figures", exist_ok=True)

# Style settings (IEEE-friendly)
plt.rcParams.update({
    "font.size": 11,
    "font.family": "serif",
    "axes.labelsize": 12,
    "axes.titlesize": 13,
    "legend.fontsize": 10,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "figure.dpi": 300,
    "savefig.dpi": 300,
    "savefig.bbox": "tight"
})

# ------------------------------------------------------------
# Common data (estimated results — replace with real values)
# ------------------------------------------------------------
models = ["LogReg", "RF", "XGB", "MLP", "1D-CNN", "LSTM", "GRU", "GCN", "TGNN"]
accuracy_bin = [74.1, 93.4, 94.7, 91.2, 92.3, 94.1, 94.3, 95.1, 96.8]
f1_bin       = [72.1, 92.4, 93.8, 90.2, 91.3, 93.1, 93.3, 94.1, 95.9]
precision_bin= [72.8, 92.8, 94.1, 90.6, 91.7, 93.5, 93.7, 94.5, 96.2]
recall_bin   = [71.5, 92.1, 93.5, 89.8, 91.0, 92.8, 93.0, 93.8, 95.6]

accuracy_mc  = [51.2, 71.2, 73.4, 67.8, 70.2, 76.1, 76.8, 78.2, 82.1]
macro_f1_mc  = [34.8, 56.8, 59.2, 52.8, 55.8, 62.8, 63.5, 65.8, 70.8]
weighted_f1_mc = [56.1, 74.2, 76.4, 71.2, 73.6, 78.9, 79.5, 81.2, 84.7]
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

x = np.arange(len(models))
width = 0.35

# --- (a) Accuracy vs F1 ---
axes[0].bar(x - width/2, accuracy_bin, width, label="Accuracy",
            color="steelblue", edgecolor="black")
axes[0].bar(x + width/2, f1_bin, width, label="F1-Score",
            color="coral", edgecolor="black")
axes[0].set_ylabel("Score (%)")
axes[0].set_xlabel("Model")
axes[0].set_title("(a) Binary Classification: Accuracy vs F1")
axes[0].set_xticks(x)
axes[0].set_xticklabels(models, rotation=30, ha="right")
axes[0].set_ylim(60, 100)
axes[0].legend(loc="lower right")
axes[0].grid(axis="y", alpha=0.3)

# Highlight TGNN
axes[0].add_patch(Rectangle((x[-1]-width, 60), 2*width, 40,
                             fill=False, edgecolor="red", lw=2))

# --- (b) Precision vs Recall ---
axes[1].bar(x - width/2, precision_bin, width, label="Precision",
            color="mediumseagreen", edgecolor="black")
axes[1].bar(x + width/2, recall_bin, width, label="Recall",
            color="gold", edgecolor="black")
axes[1].set_ylabel("Score (%)")
axes[1].set_xlabel("Model")
axes[1].set_title("(b) Binary Classification: Precision vs Recall")
axes[1].set_xticks(x)
axes[1].set_xticklabels(models, rotation=30, ha="right")
axes[1].set_ylim(60, 100)
axes[1].legend(loc="lower right")
axes[1].grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.savefig("figures/fig1_model_comparison_binary.png")
plt.savefig("figures/fig1_model_comparison_binary.pdf")
plt.show()
print("✅ Figure 1 saved")
```

```python

fig, ax = plt.subplots(figsize=(10, 5))

x = np.arange(len(models))
width = 0.27

ax.bar(x - width, accuracy_mc, width, label="Accuracy",
       color="steelblue", edgecolor="black")
ax.bar(x, macro_f1_mc, width, label="Macro-F1",
       color="coral", edgecolor="black")
ax.bar(x + width, weighted_f1_mc, width, label="Weighted-F1",
       color="mediumseagreen", edgecolor="black")

ax.set_ylabel("Score (%)")
ax.set_xlabel("Model")
ax.set_title("Multi-Class Classification Performance (15 Classes)")
ax.set_xticks(x)
ax.set_xticklabels(models, rotation=30, ha="right")
ax.set_ylim(0, 100)
ax.legend(loc="upper left")
ax.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.savefig("figures/fig2_multiclass_performance.png")
plt.savefig("figures/fig2_multiclass_performance.pdf")
plt.show()
print("✅ Figure 2 saved")
```

```python
classes = [
    "Benign", "DoS", "DDoS", "Blackhole", "Wormhole", "Replay",
    "BH+DDoS", "DDoS+WH", "BH+WH", "DDoS+RP",
    "DoS+WH", "BH+DoS", "BH+RP", "RP+WH", "DoS+RP"
]
f1_perclass = [
    97.8, 91.1, 89.9, 67.8, 61.8, 53.6,
    71.2, 66.3, 61.2, 56.6,
    55.1, 59.8, 50.1, 47.6, 51.7
]

colors = ["#2ecc71"] + ["#e74c3c"]*2 + ["#f39c12"]*3 + ["#9b59b6"]*9

fig, ax = plt.subplots(figsize=(12, 5))
bars = ax.bar(classes, f1_perclass, color=colors, edgecolor="black")

# Value labels
for b, v in zip(bars, f1_perclass):
    ax.text(b.get_x() + b.get_width()/2, v + 0.5, f"{v:.1f}",
            ha="center", fontsize=8)

ax.set_ylabel("F1-Score (%)")
ax.set_xlabel("Attack Class")
ax.set_title("Per-Class F1-Score (TGNN Model)")
ax.set_ylim(0, 105)
ax.axhline(y=70, color="gray", linestyle="--", alpha=0.5)
plt.xticks(rotation=35, ha="right")
ax.grid(axis="y", alpha=0.3)

# Legend
from matplotlib.patches import Patch
legend_elems = [
    Patch(facecolor="#2ecc71", label="Benign"),
    Patch(facecolor="#e74c3c", label="High-rate attacks"),
    Patch(facecolor="#f39c12", label="Stealth attacks"),
    Patch(facecolor="#9b59b6", label="Collaborative attacks"),
]
ax.legend(handles=legend_elems, loc="upper right")

plt.tight_layout()
plt.savefig("figures/fig3_perclass_f1.png")
plt.savefig("figures/fig3_perclass_f1.pdf")
plt.show()
print("✅ Figure 3 saved")
```

```python
scenarios = ["LogReg", "RF", "XGB", "LSTM", "GRU", "GCN", "TGNN"]
in_domain  = [72.1, 92.4, 93.8, 93.1, 93.3, 94.1, 95.9]
cross_domain = [52.8, 71.2, 73.4, 77.6, 78.2, 81.2, 89.2]
drop = [(c - i) for i, c in zip(in_domain, cross_domain)]

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

x = np.arange(len(scenarios))
width = 0.35

# --- (a) In-domain vs Cross-domain ---
axes[0].bar(x - width/2, in_domain, width, label="In-Domain F1",
            color="steelblue", edgecolor="black")
axes[0].bar(x + width/2, cross_domain, width, label="Cross-Domain F1",
            color="coral", edgecolor="black")
axes[0].set_ylabel("F1-Score (%)")
axes[0].set_xlabel("Model")
axes[0].set_title("(a) In-Domain vs Cross-Domain F1")
axes[0].set_xticks(x)
axes[0].set_xticklabels(scenarios, rotation=30, ha="right")
axes[0].set_ylim(0, 105)
axes[0].legend()
axes[0].grid(axis="y", alpha=0.3)

# --- (b) Drop ---
colors2 = ["#e74c3c"]*5 + ["#f39c12", "#2ecc71"]
axes[1].bar(scenarios, drop, color=colors2, edgecolor="black")
for i, v in enumerate(drop):
    axes[1].text(i, v - 1, f"{v:.1f}", ha="center", fontsize=9, color="white")
axes[1].set_ylabel("F1 Drop (%)")
axes[1].set_xlabel("Model")
axes[1].set_title("(b) Cross-Condition Drop (5→20 drones)")
axes[1].set_xticklabels(scenarios, rotation=30, ha="right")
axes[1].set_ylim(-30, 0)
axes[1].grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.savefig("figures/fig4_cross_condition.png")
plt.savefig("figures/fig4_cross_condition.pdf")
plt.show()
print("✅ Figure 4 saved")
```

```python
configs = [
    "TGNN (full)", "w/o Temporal\nAttention", "w/o Graph\nConvolution",
    "w/o Edge\nFeatures", "w/o Node\nFeatures", "GCN only\n(no temporal)",
    "LSTM only\n(no graph)"
]
f1_ablation = [95.9, 93.2, 91.2, 92.4, 89.8, 94.1, 93.1]
delta = [0, -2.7, -4.7, -3.5, -6.1, -1.8, -2.8]

fig, ax = plt.subplots(figsize=(11, 5))
colors = ["#2ecc71"] + ["#e74c3c"]*4 + ["#f39c12"]*2
bars = ax.barh(configs, f1_ablation, color=colors, edgecolor="black")

for b, v, d in zip(bars, f1_ablation, delta):
    label = f"{v:.1f}%" if d == 0 else f"{v:.1f}% ({d:+.1f})"
    ax.text(v + 0.3, b.get_y() + b.get_height()/2, label,
            va="center", fontsize=10)

ax.set_xlabel("F1-Score (%)")
ax.set_title("Ablation Study: Contribution of TGNN Components")
ax.set_xlim(80, 100)
ax.grid(axis="x", alpha=0.3)
ax.invert_yaxis()

plt.tight_layout()
plt.savefig("figures/fig5_ablation.png")
plt.savefig("figures/fig5_ablation.pdf")
plt.show()
print("✅ Figure 5 saved")
```

```python
horizons = [10, 30, 60, 120]
tgnn_f1   = [95.9, 92.1, 87.2, 79.8]
xgb_f1    = [93.8, 89.2, 83.2, 74.8]
rf_f1     = [92.4, 87.8, 81.2, 71.8]
lstm_f1   = [93.1, 88.5, 82.1, 73.1]

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(horizons, tgnn_f1, "o-", label="TGNN (Ours)",
        color="#2ecc71", linewidth=2.5, markersize=9)
ax.plot(horizons, xgb_f1, "s--", label="XGBoost",
        color="#e74c3c", linewidth=2, markersize=8)
ax.plot(horizons, lstm_f1, "^--", label="LSTM",
        color="#f39c12", linewidth=2, markersize=8)
ax.plot(horizons, rf_f1, "d--", label="Random Forest",
        color="#3498db", linewidth=2, markersize=8)

ax.set_xlabel("Prediction Horizon (seconds)")
ax.set_ylabel("F1-Score (%)")
ax.set_title("Prediction Horizon vs F1-Score")
ax.set_xticks(horizons)
ax.set_ylim(65, 100)
ax.legend(loc="lower left")
ax.grid(alpha=0.3)

plt.tight_layout()
plt.savefig("figures/fig6_horizon.png")
plt.savefig("figures/fig6_horizon.pdf")
plt.show()
print("✅ Figure 6 saved")
```
