```python
for i in range(10):
  #printing 0 to 9 and
  #only one number in side () means stoping number
  print (i)
```

```python
for i in range(2,10):
  #printing 0 to 9 and
  #only 2 number in side () means starting and  stoping number
  print (i)
```

```python
for i in range(2,11,2):
  #printing 0 to 9 and
  #only 3 number in side () means  starting ,stoping and step number
  print (i)
```

```python
for i in range(100,0 ,-10):
  print(i)
```
