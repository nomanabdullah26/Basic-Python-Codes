# Dictionary

dictionary are used to store data value in key:value pairs .
They are unordered , mutable and dont allow duplicate keys

```python
info ={

       "Name": " MD ABDULLAH AL NOMAN",
       "ID": "22-47155-1",
       "AGE ": 26,
       "Courses": ["Nlp","ml","cvpr","bgt"],
       "Is adult": True
}
print(info)
print(len(info))
print(type(info))
print(info["Name"])
print(info["Courses"])
print(info["Courses"][2])

info["Name"]="Noman"
print(info)

info["name"]="Noman"
print(info)

null_dic ={}
print(null_dic)
```

METHODS OF DICTIONARY

```python
#nested dictionary
student={

         "name": "rahim",
         "subject":{"phy": 97,
                      "che": 89,
                      "math": 90}
}

print(student)
print(student["subject"]["che"])

print(student.keys())
print(list(student.keys()))

print(student.values())
print(list(student.values()))

print(type(student.items()))

student.update ({"name": "MD RAHIM","city": "Sylhet"})
print(student)
```
