```python
#wap to fnd  the sum of first n number (using for loop)
num = 5

sum= 0
for i in range(1 , num+1):
  sum += i


print ("total sum = " ,sum)
```

```python
#wap to fnd  the sum of first n number (using while loop)
n=7
sum = 0
i =1

while i<=n:
  sum +=i
  i+=1

print("total sum = " , sum)
```
