# Tules

A builtin dataType that let us *immutable* sequence of value

```python
marks= (40,50,70,60,40,30,20)
print(marks)
print (marks[2])

print(marks[1:4])
print(marks[:4])
print(marks[1:])
print(marks[:])
print(marks[-1:-4])
```

```python
tup =(56)
print(type(tup))

tup =("abd")
print(type(tup))

tup =('a')
print(type(tup))

tup =(12.3)
print(type(tup))

tup =(12.3333333)
print(type(tup))

```
