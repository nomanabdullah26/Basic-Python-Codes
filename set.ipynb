# Set

Set is a  collectuion of Unordered item. Each element in the set must be unique and immutable.list and Dict are not excepted in set

```python
set_work={2,3,4,5,6,3,2,1,3}
print(set_work)
print(type(set_work))
print(len(set_work))
```

```python
info={}
print(type(info))


info=set()
print(type(info))
```

Methods of set

```python
information= set()

#add function
information.add(1)
information.add(2)
information.add(3)
information.add(4)
information.add(5)

print(information)


information.remove(3)
print(information)

information.clear()
print(information)
```

```python
set_1= {1,2,3,4,5,6}
set_2={1,3,5,7,9}

print(set_1.union(set_2))
print(set_1.intersection(set_2))
```
