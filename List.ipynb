# List
builtin data type that store set of values .

In Python : string are Immutable , but List are Mutable

```python
Marks = [45.4, 75.5, 40 ,60, 70, 90 ]
print (Marks)
print (type(Marks))

print(Marks[0])
Marks[3]= "Book"
print (Marks)

Marks.reverse()
print (Marks)

```

# Slicing

```python
Mark= [40,59,38,59,44,35,34,60]

print(Mark[1:4])
print(Mark[:4])
print(Mark[1:])
print(Mark[:])
print(Mark[-1:-4])
```

# List Methods

```python
Mark= [40,59,38,59,44,35,34,60]

Mark.append(70)#add something
print (Mark)

print(Mark.sort())#it will return 0
print (Mark)

Mark.sort(reverse=True)
print (Mark)


Mark.sort(reverse=False)
print (Mark)

Mark.reverse()
print (Mark)

Mark.insert(2,100)
print(Mark)

Mark.remove(100)
print(Mark)

Mark.pop(4)
print(Mark)
```
