```python
#Escape sequence
str1= "this is noman \n lets go"
print(str1)
```

```python
#concatenation

str1= "Bangladesh ia a country of river."
print (len(str1))

str2="some of the rivers are very dengerios"
print (len(str2))


str_final = str1+ str2
print(str_final)
print (len(str_final))
```

```python
str1= "bangladesh ia a country of river"
print(str1[7])
#slicing
print (str1[3:9])#3 to 8
print (str1[:9])#o to 9
print (str1[3:])#0 to end
print (str1[3:len(str1)])#0 to end

#negative index
print (str1[-5:-1])

#endswith
print (str1.endswith("River"))

#capitalize
print (str1.capitalize())

#replace
print (str1.replace("b","a"))

#find
print (str1.find("book"))

#count
print (str1.count("a"))
```

Example of If else condition

```python
mark= int(input("Enter Your computer science makrk : "))

if (mark>=90):
    print(" your result is A+")

elif (mark>=85):
    print(" your result is A")

elif (mark>=80):
    print(" your result is B+")

elif (mark>=75):
    print(" your result is B")

elif (mark>=70):
    print(" your result is C+")

elif (mark>=65):
    print(" your result is C")

elif (mark>=60):
    print(" your result is D+")

elif (mark>=55):
    print(" your result is D")

else:
  print(" FAIL!!")

```
